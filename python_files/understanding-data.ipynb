## b)Pie Chart

```python
import matplotlib.pyplot as plt

counts = df["Survived"].value_counts()
print(counts)
plt.pie(counts, labels=counts.index, autopct="%1.1f%%")
plt.show()
```

```python
import matplotlib.pyplot as plt

counts = df["Pclass"].value_counts()
print(counts)
plt.pie(counts, labels=counts.index, autopct="%1.1f%%")
plt.show()
```

```python
name = []
rating = []
reviews = []
salary = []
interviews = []
jobs = []
benefits = []
photos = []
company_type = []
location = []
highly_rated_for = []
critically_rated_for = []

for i in company:

    # Company Name
    name.append(
        i.find('h2', class_='companyCardWrapper__companyName').text.strip()
    )

    # Rating
    rating.append(
        i.find('div', class_='companyCardWrapper__companyRating').text.strip()
    )

    # Company Type and Location
    info = i.find(
        'span',
        class_='companyCardWrapper__interLinking'
    )

    if info:
        info = info.text.strip()

        parts = info.split('|')

        company_type.append(parts[0].strip())

        if len(parts) > 1:
            location.append(parts[1].strip())
        else:
            location.append(None)
    else:
        company_type.append(None)
        location.append(None)

    # Highly Rated / Critically Rated
    rating_values = i.find_all(
        'span',
        class_='companyCardWrapper__ratingValues'
    )

    if len(rating_values) > 0:
        highly_rated_for.append(
            rating_values[0].text.strip()
        )
    else:
        highly_rated_for.append(None)

    if len(rating_values) > 1:
        critically_rated_for.append(
            rating_values[1].text.strip()
        )
    else:
        critically_rated_for.append(None)

    # Reviews, Salary, Interviews, Jobs, Benefits, Photos
    actions = i.find_all(
        'a',
        class_='companyCardWrapper__ActionWrapper'
    )

    counts = []

    for action in actions:
        count = action.find(
            'span',
            class_='companyCardWrapper__ActionCount'
        )

        if count:
            counts.append(count.text.strip())
        else:
            counts.append(None)

    # Get each value safely
    reviews.append(counts[0] if len(counts) > 0 else None)
    salary.append(counts[1] if len(counts) > 1 else None)
    interviews.append(counts[2] if len(counts) > 2 else None)
    jobs.append(counts[3] if len(counts) > 3 else None)
    benefits.append(counts[4] if len(counts) > 4 else None)
    photos.append(counts[5] if len(counts) > 5 else None)


# Create DataFrame
df = pd.DataFrame({
    'name': name,
    'rating': rating,
    'reviews': reviews,
    'salary': salary,
    'interviews': interviews,
    'jobs': jobs,
    'benefits': benefits,
    'photos': photos,
    'company_type': company_type,
    'location': location,
    'highly_rated_for': highly_rated_for,
    'critically_rated_for': critically_rated_for
})

df.head()
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/139.0.0.0 Safari/537.36"
}

name = []
rating = []
reviews = []
salary = []
interviews = []
jobs = []
benefits = []
photos = []
company_type = []
location = []
highly_rated_for = []
critically_rated_for = []

for page in range(1, 11):

    url = f'https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav&page={page}'

    webpage = requests.get(url, headers=headers)

    soup = BeautifulSoup(webpage.text, 'html.parser')

    company = soup.find_all(
        'div',
        class_='companyCardWrapper'
    )

    print("Page:", page, "Companies:", len(company))

    for i in company:

        # Name
        x = i.find('h2', class_='companyCardWrapper__companyName')
        name.append(x.text.strip() if x else None)

        # Rating
        x = i.find('div', class_='companyCardWrapper__companyRating')
        rating.append(x.text.strip() if x else None)

        # Company type and location
        x = i.find('span', class_='companyCardWrapper__interLinking')

        if x:
            info = x.text.strip()
            parts = info.split('|')

            company_type.append(parts[0].strip())

            if len(parts) > 1:
                location.append(parts[1].strip())
            else:
                location.append(None)
        else:
            company_type.append(None)
            location.append(None)

        # Highly / Critically rated
        rating_values = i.find_all(
            'span',
            class_='companyCardWrapper__ratingValues'
        )

        if len(rating_values) > 0:
            highly_rated_for.append(
                rating_values[0].text.strip()
            )
        else:
            highly_rated_for.append(None)

        if len(rating_values) > 1:
            critically_rated_for.append(
                rating_values[1].text.strip()
            )
        else:
            critically_rated_for.append(None)

        # Reviews, Salary, Interviews, Jobs, Benefits, Photos
        actions = i.find_all(
            'a',
            class_='companyCardWrapper__ActionWrapper'
        )

        counts = []

        for action in actions:

            x = action.find(
                'span',
                class_='companyCardWrapper__ActionCount'
            )

            counts.append(x.text.strip() if x else None)

        reviews.append(counts[0] if len(counts) > 0 else None)
        salary.append(counts[1] if len(counts) > 1 else None)
        interviews.append(counts[2] if len(counts) > 2 else None)
        jobs.append(counts[3] if len(counts) > 3 else None)
        benefits.append(counts[4] if len(counts) > 4 else None)
        photos.append(counts[5] if len(counts) > 5 else None)
```
